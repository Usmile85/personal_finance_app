```python
import pandas as pd

# Load your existing file
df = pd.read_csv("personal_finance_sample.csv")

# Convert 'Date' to datetime
df["Date"] = pd.to_datetime(df["Date"], errors="coerce", dayfirst=True)

# Drop any rows with invalid dates
df = df.dropna(subset=["Date"])

# Save to new CSV
df.to_csv("personal_finance_cleaned.csv", index=False)

print("✅ Cleaned data saved as personal_finance_cleaned.csv")
```
